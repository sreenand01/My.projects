```python
# replacing null

df['danceability']=df['danceability'].fillna(df['danceability'].mean())

df['speechiness']=df['speechiness'].fillna(df['speechiness'].mean())

df['valence']=df['valence'].fillna(df['valence'].mean())
```

```python
df['artists']=df['artists'].fillna(df['artists'].mode()[0])

df['album_name']=df['album_name'].fillna(df['album_name'].mode()[0])

df['track_name']=df['track_name'].fillna(df['track_name'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.drop('Unnamed: 0',axis=1,inplace=True)
```

```python
# Heatmap
numbers = df.select_dtypes(include='number')

plt.figure(figsize=(14, 10))

sns.heatmap(
    numbers.corr(),
    cmap='coolwarm',
    annot=True,
    fmt='.2f'
)
```

```python
# Box plot
numbers = df.select_dtypes(include='number')

for column in numbers.columns:
    plt.figure(figsize=(6, 4))
    sns.boxplot(y=df[column], color='#1DB954')
    plt.title("Box Plot of " + column)
    plt.ylabel(column)
    plt.show()
```

```python
for col in df.select_dtypes(include=['int','float']).columns.tolist():
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower,upper)
    # outlier = ((df[col] < lower) | (df[col] > upper))
    # df.loc[outlier,col] = df[col].median()
```

```python
numbers = df.select_dtypes(include='number')

for column in numbers.columns:
    plt.figure(figsize=(6, 4))
    sns.boxplot(y=df[column], color='#1DB954')
    plt.title("Box Plot of " + column)
    plt.ylabel(column)
    plt.show()
```

```python
#Histplot
sns.histplot(data=df, x='popularity', bins=20, color='green', edgecolor='black')

plt.title("Distribution of Track Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Tracks")
plt.show()
```

```python
# Count plot: Top 10 genres popularity >50
top_genres = df['track_genre'].value_counts().head(10).index
popular_df = df[(df['track_genre'].isin(top_genres)) & (df['popularity'] > 50)]

sns.countplot(data=popular_df, y='track_genre', order=top_genres, color='#1DB954', edgecolor='black')
plt.title("Number of Popular Tracks by Genre")
plt.xlabel("Count of Popular Tracks")
plt.ylabel("Genre")
plt.show()
```

```python
# Bar chartTop 10 Artists
df['artists'].value_counts().head(10).plot(
    kind='bar',
    color='#1DB954',
    edgecolor='black'
)
plt.title("Top 10 Artists by Track Count")
plt.xlabel("Artist")
plt.ylabel("Number of Tracks")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Bar plot: Average popularity by genre
df.groupby('track_genre')['popularity'].mean().nlargest(10).plot(kind='bar', color='#1DB954', edgecolor='black')
plt.title("Top 10 Genres by Average Popularity")
plt.ylabel("Average Popularity")
plt.xticks(rotation=45)
plt.show()
```

```python
# Scatter plot
sns.scatterplot(data=df, x='tempo', y='loudness', color='#1DB954')

plt.title("Tempo vs Loudness")
plt.xlabel("Tempo (BPM)")
plt.ylabel("Loudness (dB)")
plt.show()
```

```python
# Line plot: Popularity by track duration
df['duration_minutes'] = df['duration_ms'] / 60000
df['duration_group'] = pd.cut(df['duration_minutes'], bins=10)

df.groupby('duration_group', observed=True)['popularity'].mean().plot(kind='line', color='#1DB954', marker='o')
plt.title("Average Popularity by Track Duration")
plt.xlabel("Track Duration Groups")
plt.ylabel("Average Popularity")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Pie chart: Explicit vs Non-Explicit Tracks
df['explicit'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=['#1DB954', '#B3EFC5']
)
plt.title("Explicit vs Non-Explicit Tracks")
plt.ylabel('')
plt.show()
```
